```python
from deepeval import evaluate
from deepeval.test_case import LLMTestCase
from deepeval.metrics import BaseMetric

# --- 1. Create a Dummy Metric (No LLM needed) ---
class DummyMetric(BaseMetric):
    def __init__(self):
        self.threshold = 0.5
        self.name = "Dummy Metric"

    def measure(self, test_case: LLMTestCase):
        self.score = 1.0  # Hardcoded success
        self.reason = "This is a dummy reason for inspection."
        self.success = self.score >= self.threshold
        return self.score

    async def a_measure(self, test_case: LLMTestCase):
        return self.measure(test_case)

    def is_successful(self):
        return self.success

    @property
    def __name__(self):
        return "Dummy Metric"

# --- 2. Create One Test Case ---
test_case = LLMTestCase(
    input="Sample Input",
    actual_output="Sample Output",
    retrieval_context=["Chunk 1", "Chunk 2"]
)

# --- 3. Run Evaluate ---
# print_results=False hides the default table so we can print the object raw
results = evaluate([test_case], [DummyMetric()])
print(results.test_results[0].metrics_data)
```
